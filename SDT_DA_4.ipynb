```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=pd.read_csv("C:/Users/Kota Mounika/Downloads/marketing_campaign.csv",sep=';')
data.head()
```

```python
data.info()
```

```python
data.describe()
```

```python
data.isnull().sum()
```

```python
#filling missing income values
data['Income']=data['Income'].fillna(data['Income'].median())
```

```python
data.isnull().sum()
```

```python
#New column Total Spending
data['Total_Spending']=(data['MntWines']+data['MntFruits']+data['MntMeatProducts']+data['MntFishProducts']+data['MntGoldProds'])
```

```python
data.columns
```

```python
data=data.dropna()
```

```python
data=data.drop_duplicates()
```

```python
#Age Distribution
plt.figure(figsize=(8,5))
sns.histplot(data['Year_Birth'],bins=30,kde=True)
plt.title('Custom Birth Year Distribution')
plt.show()
```

```python
#Income vs Spending
plt.figure(figsize=(8,5))
sns.scatterplot(x='Income',y='Total_Spending',data=data)
plt.title('Income vs Total Spending')
plt.show()
```

```python
#Marketing channel Performance
channels=['NumWebPurchases','NumCatalogPurchases','NumStorePurchases']
channel_sales=data[channels].sum()
plt.figure(figsize=(8,5))
channel_sales.plot(kind='bar')
plt.title('Purchases by Marketing Channel')
plt.ylabel('Total Purchases')
plt.show()
```

```python
#campaign acceptance
campaigns=['AcceptedCmp1', 'AcceptedCmp2', 'AcceptedCmp3', 'AcceptedCmp4',
       'AcceptedCmp5']
campaign_data=data[campaigns].sum()
plt.figure(figsize=(8,5))
campaign_data.plot(kind='bar',color='orange')
plt.title("Campaign Acceptance")
plt.ylabel('Accepted count')
plt.show()
```

```python
#correlation heatmap
plt.figure(figsize=(12,8))
sns.heatmap(data.corr(numeric_only=True),cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

```python
filtered_data = data[data['Income'] < 120000]

plt.figure(figsize=(8,5))
sns.scatterplot(x='Income', y='Total_Spending', data=filtered_data)

plt.title('Income vs Total Spending')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.regplot(
    x='Income',
    y='Total_Spending',
    data=filtered_data,
    scatter_kws={'alpha':0.5}
)

plt.title('Income vs Total Spending')
plt.show()
```
